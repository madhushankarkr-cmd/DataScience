```python
df.loc[5, "Cylinders"] = np.nan
df.loc[10, "Seating_Capacity"] = np.nan
```

```python
df.isnull().sum()
```

```python
df["Cylinders"] = df["Cylinders"].fillna(
    df["Cylinders"].mean()
)
```

```python
df["Seating_Capacity"] = df["Seating_Capacity"].fillna(
    df["Seating_Capacity"].median()
)
```

```python
df["Fuel_Type"] = df["Fuel_Type"].fillna(
    df["Fuel_Type"].mode()[0]
)
```
