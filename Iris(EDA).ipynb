# **Check Missing Value**

```python
df.isnull().sum()
```

```python
df.loc[5, "SepalLengthCm"] = np.nan
df.loc[10, "SepalWidthCm"] = np.nan
df.loc[15, "PetalLengthCm"] = np.nan
df.loc[20, "PetalWidthCm"] = np.nan
```

```python
df.isnull().sum()
```

```python
df["SepalLengthCm"] = df["SepalLengthCm"].fillna(
    df["SepalLengthCm"].mean()
)
```

```python
df["SepalWidthCm"] = df["SepalWidthCm"].fillna(
    df["SepalWidthCm"].median()
)
```

```python
df["PetalLengthCm"] = df["PetalLengthCm"].fillna(
    df["PetalLengthCm"].mode()[0]
)
```

```python
df["PetalWidthCm"] = df["PetalWidthCm"].fillna(
    df["PetalWidthCm"].mean()
)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df["Species"].value_counts()
```

```python
df.hist(figsize=(10, 8))
plt.show()
```

```python
sns.boxplot(data=df)

plt.show()
```

```python
sns.countplot(x="Species", data=df)

plt.show()
```

# **Sepal Length vs Sepal Width**

```python
sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="SepalWidthCm",
    hue="Species"
)

plt.show()
```

# Petal Length vs Petal **Width**

```python
sns.scatterplot(
    data=df,
    x="PetalLengthCm",
    y="PetalWidthCm",
    hue="Species"
)

plt.show()
```
