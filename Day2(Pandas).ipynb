# **Missing value Treatment**

```python
df2 = pd.DataFrame({
    "S_name": ["Aman", "Riya","Madhu","Gaurav","ALex","Peanut","rohit","Virat","Thala","Rahul"],
    "S_Age": [18,17,np.nan,17,18,17,np.nan,18,np.nan,18],
    "S_marks": [400,np.nan,470,np.nan,460,np.nan,410,430,np.nan,440],
    "S_department": ["CSE","EE",np.nan,"ECSE",np.nan,"EE","AIDS",np.nan,"CSE",np.nan]
})
df2
```

```python
#TO show missingfunction
df2.isnull()
```

```python
df2.isnull().sum()
```

```python
#to delete row
df2.dropna()
```

```python
#to  delete column
df2.dropna(axis=1)
```

```python
S_Age_mean = df2['S_Age'].mean()
print(int(S_Age_mean))
```

```python
df2['S_Age']=df2['S_Age'].fillna(df2['S_Age'].mean())
df2
```

```python
df2['S_marks']=df2['S_marks'].fillna(df2['S_marks'].median())
df2
```

```python
df2['S_department']=df2['S_department'].fillna(df2['S_department'].mode()[0])
df2
```
