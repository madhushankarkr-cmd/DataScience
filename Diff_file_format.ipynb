# **SQLite**

```python
import sqlite3
```

```python
conn=sqlite3.connect("studentrecord.db")
cursor=conn.cursor()
print("DataBase created Successfully")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS result(S_ID INTEGER,S_Name CHAR,Marks INTEGER)
""")
print("Table creted successfully")
```

```python
cursor.execute("INSERT INTO result VALUES(?,?,?)",
               (1,"Alex",99))
cursor.execute("INSERT INTO result VALUES(?,?,?)",
               (1,"Anya",99))
conn.commit()
print("Record Inserted")
conn.close()
```

```python
conn=sqlite3.connect("studentrecord.db")
cursor=conn.cursor()
cursor.execute("SELECT *FROM result")
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
```
