```python
arr = np.array([10, 20, 30, 40, 50])

# View
view = arr[1:4]
view[0] = 999

print("Original after view:", arr)

# Copy
copy_arr = arr[1:4].copy()
copy_arr[0] = 100

print("Original after copy:", arr)
print("Copy:", copy_arr)
```

```python
arr = np.arange(24).reshape(2, 3, 4)

print("Original:")
print(arr)

print("Transpose:")
print(arr.transpose(2, 1, 0))

print("Swap axes:")
print(arr.swapaxes(0, 2))
```
